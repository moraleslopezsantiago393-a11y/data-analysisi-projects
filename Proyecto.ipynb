```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


drivers = pd.read_csv('drivers.csv')
results = pd.read_csv('results.csv')


df_combined = pd.merge(results, drivers, on='driverId')


plt.figure(figsize=(10, 6))
sns.countplot(data=df_combined, y='nationality', 
              order=df_combined['nationality'].value_counts().index[:15])

plt.title('Top 15 Nacionalidades en F1')
plt.show()
```
